5.  What are dunder methods in Python?
  
  In Python, dunder methods (short for double underscore methods) are special, built-in methods that start and end with double underscores, like __init__, __str__, __len__, etc.

  They're also called magic methods because they allow you to define how your objects behave with built-in Python operations like printing, addition, comparison, and more.
  Dunder Method	   Purpose
__init__(self)	   Constructor; initializes an object
__str__(self)	     Defines what print(object) returns
__repr__(self)	   Official string representation (used in debugging)
__len__(self)	     Defines behavior for len(object)
__add__(self, other)	Defines behavior for + operator
__eq__(self, other)	  Defines behavior for == comparison
__getitem__(self, key)	Enables indexing like obj[key]


```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

book = Book("1984", 328)
print(book)        # Output: Book: 1984
print(len(book))
```

6. Explain the concept of inheritance in OOP?

  Inheritance allows a class (child/subclass) to inherit the properties (attributes) and behaviors (methods) of another class (parent/superclass).

  This means you can create new classes based on existing ones without rewriting code.

  - Promotes code reusability

  - Makes code more organized and maintainable

  - Helps create a hierarchical classification.

 eg: classes like Dog and Cat can inherit from Animal, and they’ll automatically have those attributes and methods — plus they can add their own unique behaviors, like bark() or meow()





```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

# Child class inherits from Animal
class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks.")

class Cat(Animal):
    def speak(self):
        print(f"{self.name} meows.")

dog = Dog("Buddy")
cat = Cat("Whiskers")

dog.speak()   # Output: Buddy barks.
cat.speak()
```

Types of Inheritance (Python supports all):

1. Single Inheritance: One child inherits from one parent.

2. Multiple Inheritance: One child inherits from multiple parents.

3. Multilevel Inheritance: A child inherits from a class that also inherits from another.

4. Hierarchical Inheritance: Multiple children inherit from one parent.

7. What is polymorphism in OOP?

  Polymorphism is a core concept in OOP that allows the same method or function name to behave differently based on the object that calls it.

  polymorphism = many forms.

  It allows objects of different classes to be treated as objects of a common parent class, especially when using methods that are overridden differently in each subclass.




```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Polymorphism in action
def animal_sound(animal):
    animal.speak()

# Different behaviors with same interface
dog = Dog()
cat = Cat()

animal_sound(dog)  # Output: Dog barks
animal_sound(cat)  # Output: Cat meows
```


*   speak() is the same method name, but behaves differently for Dog and Cat

*   animal_sound() works with any object that has a speak() method — that’s the power of polymorphism


9. How is encapsulation achieved in Python?

   In Python, encapsulation is achieved by restricting direct access to class variables and methods, typically using access modifiers and getter/setter methods.

   Encapsulation = binding data and methods into one unit (a class) and protecting that data from being accessed or modified directly from outside the class.
   
   1. Access Modifiers:
   Python doesn’t have traditional private/protected keywords like Java or C++, but it uses naming conventions:

   Modifier	  Syntax	Access Level

   Public	    var	    Accessible from anywhere

   Protected	_var	  Convention: internal use only (can still access  
                      from outside)
                      
   Private	 __var	    Name mangled; not directly accessible from outside


```python
class Person:
    def __init__(self, name, age):
        self.__name = name     # private attribute
        self.__age = age       # private attribute

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, new_name):
        self.__name = new_name

    # Getter for age
    def get_age(self):
        return self.__age

    # Setter for age
    def set_age(self, new_age):
        if new_age > 0:
            self.__age = new_age
        else:
            print("Age must be positive.")
p = Person("Alice", 30)

print(p.get_name())  # ✅ Allowed
p.set_age(35)        # ✅ Allowed

print(p.__name)
```

__name becomes _Person__name internally — this is called name mangling.

It’s not truly private (you can access it using p._Person__name), but it’s a strong convention to prevent accidental access.

11.  What is method overloading in Python?

    Method overloading means having multiple methods in the same class with the same name but different parameters (number or type).

*  It’s a way to define multiple behaviors for a method based on how it’s called.

* Python does not support traditional method overloading natively.
  If you define a method multiple times in the same class, only the last one will be used.



```python
class Demo:
    def greet(self):
        print("Hello")

    def greet(self, name):
        print(f"Hello {name}")

# Only the second greet() survives
d = Demo()
d.greet("Alice")  # ✅ Works
# d.greet()
```

 simulate method overloading in Python:
1. Use default arguments:

```python
class Demo:
    def greet(self, name=None):
        if name:
            print(f"Hello {name}")
        else:
            print("Hello")

d = Demo()
d.greet()         # Output: Hello
d.greet("Alice")
```

13. What is a property decorator in Python?

   The @property decorator lets you define a method that can be accessed like an attribute — no need to call it with parentheses ().

   It’s commonly used to:

  Control access to private attributes

  Add logic/validation when getting or setting values

  Achieve encapsulation in a Pythonic way

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value > 0:
            self._radius = value
        else:
            raise ValueError("Radius must be positive")

    @property
    def area(self):
        from math import pi
        return pi * self._radius ** 2
c = Circle(5)
print(c.radius)     # ✔️ Access like attribute: 5
c.radius = 10       # ✔️ Setter with validation
print(c.area)
# c.radius = -2    #value error
```

* @property turns a method into a getter

* @radius.setter defines the setter for that property

* @property without a setter → read-only property

17. What is the difference between a class variable and an instance variable?

1.Instance Variable
Belongs to a specific object (instance)

18. What is multiple inheritance in Python?

Multiple Inheritance means a class can inherit from more than one parent class.

So, a child class can have multiple sources of behavior or attributes.

class Parent1:
    ...

class Parent2:
    ...

class Child(Parent1, Parent2):
    ...

```python
class Flyer:
    def fly(self):
        print("Flying in the sky")

class Swimmer:
    def swim(self):
        print("Swimming in the water")

class Duck(Flyer, Swimmer):
    pass

d = Duck()
d.fly()   # Output: Flying in the sky
d.swim()  # Output: Swimming in the water
```

The Diamond Problem

If two parent classes inherit from the same grandparent, and the child inherits both — Python might need to resolve which method to use if names clash.

Python handles this using MRO (Method Resolution Order).

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):  # Multiple Inheritance
    pass

d = D()
d.show()  # Output: B (Python looks left to right)
print(D.__mro__)  # Shows the method resolution order
```

21. What is the significance of the __del__ method in Python?

* __del__() method in Python — also known as the destructor method 💣

22. What is the difference between @staticmethod and @classmethod in Python?

 The difference between @staticmethod and @classmethod in Python comes down to what they receive as their first argument — and how they interact with the class.

1. @staticmethod:

*  Doesn’t take self or cls as the first argument

* Behaves like a regular function, just lives in a class for organizational purposes

*  Doesn’t access or modify class or instance state


```python
class MathUtils:
    @staticmethod
    def add(x, y):
        return x + y

print(MathUtils.add(3, 5))
```

 2. @classmethod

*  Takes cls as the first parameter (not self)

* Can access or modify class-level data

* Can be used to create factory methods — that return an instance of the class

```python
class Book:
    books_created = 0

    def __init__(self, title):
        self.title = title
        Book.books_created += 1

    @classmethod
    def from_string(cls, title_str):
        return cls(title_str)  # Creates a Book instance

b = Book.from_string("The Alchemist")  # Uses class method to construct
```

* Use @staticmethod when the method doesn’t need to access class or instance variables.

* Use @classmethod when you need to work with the class itself (e.g. creating instances, tracking class data, etc).

24.  What is method chaining in Python OOP?

 Method chaining is when you call multiple methods on the same object in a single line, like this:

 object.method1().method2().method3()

 Each method returns the object itself (usually self) so the next method can be called on it.
 **works**:
 To enable chaining, each method should return self at the end.

```python
class Person:
    def __init__(self, name):
        self.name = name
        self.skills = []

    def learn(self, skill):
        self.skills.append(skill)
        return self

    def show_skills(self):
        print(f"{self.name} knows: {', '.join(self.skills)}")
        return self

# Method chaining in action
p = Person("Hamza")
p.learn("Python").learn("SQL").show_skills()
```

**Usage:**
* When building builder patterns or fluent interfaces

* For cleaner, more readable code in object setup/configuration

* In data pipelines, like with Pandas (df.drop().filter().groupby())

If a method doesn’t return self, chaining will break:

```python
p.learn("Java").show_skills().learn("C++")  # works
p.show_skills().learn("C++")  # only works if show_skills returns self
```

25. What is the purpose of the __call__ method in Python?

   The __call__ method in Python is one of those cool dunder methods that can make your objects behave like functions. Yep, with __call__, your object becomes callable!

   __call__ is a special method that allows an instance of a class to be called as if it were a function.

```python
class Greeter:
    def __init__(self, name):
        self.name = name

    def __call__(self, message):
        print(f"{self.name} says: {message}")

g = Greeter("Hamza")

g("Hello, world!")  # This works because of __call__()
```

usage:

* Create function-like objects with internal state

* Build custom decorators

* Design elegant APIs and fluent interfaces

```python
class LengthChecker:
    def __init__(self, min_len):
        self.min_len = min_len

    def __call__(self, value):
        return len(value) >= self.min_len

check = LengthChecker(5)
print(check("hello"))
```

```python
#2. Write a program to create an abstract class Shape with a method area().
#Derive classes Circle and Rectangle from it and implement the area() method in both

from abc import ABC, abstractmethod
import math

# Abstract Base Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Derived class: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Derived class: Rectangle
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage
c = Circle(5)
print(f"Circle Area: {c.area():.2f}")  # Output: Circle Area: 78.54

r = Rectangle(4, 6)
print(f"Rectangle Area: {r.area()}")   # Output: Rectangle Area: 24
```

```python
#3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type.
#Derive a class Car and further derive a class ElectricCar that adds a battery attribute.
# Base class
class Vehicle:
    def __init__(self, vehicle_type):
        self.type = vehicle_type

    def show_type(self):
        print(f"Vehicle Type: {self.type}")

# Derived class
class Car(Vehicle):
    def __init__(self, vehicle_type, brand):
        super().__init__(vehicle_type)
        self.brand = brand

    def show_brand(self):
        print(f"Brand: {self.brand}")

# Further derived class
class ElectricCar(Car):
    def __init__(self, vehicle_type, brand, battery_capacity):
        super().__init__(vehicle_type, brand)
        self.battery = battery_capacity

    def show_battery(self):
        print(f"Battery Capacity: {self.battery} kWh")

# Example usage
ecar = ElectricCar("4-Wheeler", "Tesla", 100)
ecar.show_type()       # Output: Vehicle Type: 4-Wheeler
ecar.show_brand()      # Output: Brand: Tesla
ecar.show_battery()    # Output: Battery Capacity: 100 kWh
```

```python
#4. Demonstrate polymorphism by creating a base class Bird with a method fly().
#Create two derived classes Sparrow and Penguin that override the fly() method.
# Base class
class Bird:
    def fly(self):
        print("Some birds can fly.")

# Derived class 1
class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high in the sky.")

# Derived class 2
class Penguin(Bird):
    def fly(self):
        print("Penguins can't fly, they swim instead.")

# Polymorphism in action
birds = [Sparrow(), Penguin()]

for bird in birds:
    bird.fly()  # Each object responds differently to the same method call
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
# balance and methods to deposit, withdraw, and check balance.
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ₹{amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ₹{amount}")
        else:
            print("Insufficient balance or invalid amount.")

    def check_balance(self):
        print(f"Current Balance: ₹{self.__balance}")

# Example usage
account = BankAccount(1000)
account.check_balance()   # ₹1000
account.deposit(500)      # Deposited ₹500
account.withdraw(300)     # Withdrew ₹300
account.check_balance()   # ₹1200
```

```python
#6. Demonstrate runtime polymorphism using a method play() in a base class Instrument.
#Derive classes Guitar and Piano that implement their own version of play().
# Base class
class Instrument:
    def play(self):
        print("Playing an instrument...")

# Derived class 1
class Guitar(Instrument):
    def play(self):
        print("Strumming the guitar")

# Derived class 2
class Piano(Instrument):
    def play(self):
        print("Playing the piano")

# Polymorphism in action
def play_music(instrument):
    instrument.play()  # This will call the correct method at runtime

# Example usage
instruments = [Guitar(), Piano()]

for inst in instruments:
    play_music(inst)
```

```python
#7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static
# method subtract_numbers() to subtract two numbers.
class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

# Example usage
sum_result = MathOperations.add_numbers(10, 5)
print(f"Addition: {sum_result}")  # Output: 15

diff_result = MathOperations.subtract_numbers(10, 5)
print(f"Subtraction: {diff_result}")  # Output: 5
```

```python
# 8. Implement a class Person with a class method to count the total number of persons created.
class Person:
    total_persons = 0

    def __init__(self, name):
        self.name = name
        Person.total_persons += 1

    @classmethod
    def count_persons(cls):
        return cls.total_persons

# Example usage
p1 = Person("Alice")
p2 = Person("Bob")
p3 = Person("Charlie")

Person.count_persons()  # Output: Total persons created: 3
```

```python
#9. Write a class Fraction with attributes numerator and denominator.
# Override the str method to display the fraction as "numerator/denominator".
class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

# Example usage
f1 = Fraction(3, 4)
f2 = Fraction(5, 8)

print(f1)  # Output: 3/4
print(f2)  # Output: 5/8
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
#vectors.
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        # Overloading the + operator
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Example usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2  # Uses __add__()

print(v1)  # Output: Vector(2, 3)
print(v2)  # Output: Vector(4, 5)
print(v3)  # Output: Vector(6, 8)
```

```python
#12. Implement a class Student with attributes name and grades.
# Create a method average_grade() to compute the average of the grades.
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)
s1 = Student("Andrew", [85, 90, 78, 92])
print(f"{s1.name}'s average grade is: {s1.average_grade():.2f}")
```

```python
#13. Create a class Rectangle with methods set_dimensions() to set the dimensions and area() to calculate the area.
class Rectangle:
    def __init__(self):
        self.length = 0
        self.width = 0

    def set_dimensions(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Example usage
rect = Rectangle()
rect.set_dimensions(5, 10)
print(f"Area of the rectangle: {rect.area()}")  # Output: 50
```

```python
#14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
#and hourly rate. Create a derived class Manager that adds a bonus to the salary.
# Base class
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

# Derived class
class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

# Example usage
emp = Employee("John", 40, 20)
mgr = Manager("Alice", 40, 30, 500)

print(f"{emp.name}'s Salary: ₹{emp.calculate_salary()}")    # Output: ₹800
print(f"{mgr.name}'s Salary: ₹{mgr.calculate_salary()}")
```

```python
#15. Create a class Product with attributes name, price, and quantity.
#Implement a method total_price() that calculates the total price of the product.
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity

# Example usage
p1 = Product("Laptop", 50000, 2)
p2 = Product("Mouse", 800, 3)

print(f"Total price for {p1.name}: ₹{p1.total_price()}")
print(f"Total price for {p2.name}: ₹{p2.total_price()}")
```

```python
#16. Create a class Animal with an abstract method sound().
#Create two derived classes Cow and Sheep that implement the sound() method.

from abc import ABC, abstractmethod

# Abstract base class
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

# Derived class Cow
class Cow(Animal):
    def sound(self):
        return "Moo"

# Derived class Sheep
class Sheep(Animal):
    def sound(self):
        return "Baa"

# Example usage
cow = Cow()
sheep = Sheep()

print(f"Cow says: {cow.sound()}")    # Output: Moo
print(f"Sheep says: {sheep.sound()}")  # Output: Baa
```

```python
#17. Create a class Book with attributes title, author, and year_published.
# Add a method get_book_info() that returns a formatted string with the book's details.
class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published

    def get_book_info(self):
        return f"'{self.title}' by {self.author} (Published in {self.year_published})"

# Example usage
book1 = Book("To Kill a Mockingbird", "Harper Lee", 1960)
book2 = Book("1984", "George Orwell", 1949)

print(book1.get_book_info())
print(book2.get_book_info())
```

```python
#18. Create a class House with attributes address and price.
#Create a derived class Mansion that adds an attribute number_of_rooms

# Base class
class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

# Derived class
class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

    def get_details(self):
        return (f"Mansion located at {self.address}, priced at ₹{self.price}, "
                f"with {self.number_of_rooms} rooms.")

# Example usage
m1 = Mansion("Beverly Hills, LA", 500000000, 15)
print(m1.get_details())
```
